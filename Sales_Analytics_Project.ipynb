```python
import pandas as pd

df = pd.read_csv('Sales Dataset.csv')
```

```python
df.head()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python
df['Order Date'].dtype
```

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
```

```python
df[['Order Date', 'Year', 'Month']].head()
```

```python
yearly_sales = df.groupby('Year')['Amount'].sum()
yearly_sales
```

```python
monthly_sales = df.groupby('Month')['Amount'].sum()
monthly_sales
```

```python
import matplotlib.pyplot as plt
```

```python
monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Amount')
plt.show()
```

```python
yearly_sales.plot(kind='bar')

plt.title('Yearly Sales')
plt.xlabel('Year')
plt.ylabel('Total Sales Amount')
plt.show()
```

```python
category_sales = df.groupby('Category')['Amount'].sum()

print(category_sales)
```

```python
category_sales.plot(kind='bar')

plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales Amount')
plt.show()
```

```python
subcategory_sales = df.groupby('Sub-Category')['Amount'].sum()

print(subcategory_sales)
```

```python
subcategory_sales.plot(kind='bar')

plt.title('Sales by Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
profit_category = df.groupby('Category')['Profit'].sum()

print(profit_category)
```

```python
profit_category.plot(kind='bar')

plt.title('Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.show()
```

```python
profit_subcategory = df.groupby('Sub-Category')['Profit'].sum()

print(profit_subcategory)
```

```python
profit_subcategory.plot(kind='bar')

plt.title('Profit by Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.show()
```

```python
state_sales = df.groupby('State')['Amount'].sum()

print(state_sales)
```

```python
state_sales.plot(kind='bar')

plt.title('Sales by State')
plt.xlabel('State')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
state_profit = df.groupby('State')['Profit'].sum()

print(state_profit)
```

```python
state_profit.plot(kind='bar')

plt.title('Profit by State')
plt.xlabel('State')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.show()
```

```python
payment_sales = df.groupby('PaymentMode')['Amount'].sum()

print(payment_sales)
```

```python
payment_sales.plot(kind='bar')

plt.title('Sales by Payment Mode')
plt.xlabel('Payment Mode')
plt.ylabel('Total Sales Amount')
plt.show()
```

```python
customer_sales = df.groupby('CustomerName')['Amount'].sum()

print(customer_sales.sort_values(ascending=False).head(10))
```

```python
top_customers = customer_sales.sort_values(ascending=False).head(10)

top_customers.plot(kind='bar')

plt.title('Top 10 Customers by Sales')
plt.xlabel('Customer Name')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
city_sales = df.groupby('City')['Amount'].sum()

print(city_sales.sort_values(ascending=False).head(10))
```

```python
top_cities = city_sales.sort_values(ascending=False).head(10)

top_cities.plot(kind='bar')

plt.title('Top 10 Cities by Sales')
plt.xlabel('City')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
quantity_category = df.groupby('Category')['Quantity'].sum()

print(quantity_category)
```

```python
quantity_category.plot(kind='bar')

plt.title('Quantity Sold by Category')
plt.xlabel('Category')
plt.ylabel('Total Quantity')
plt.show()
```

```python
quantity_subcategory = df.groupby('Sub-Category')['Quantity'].sum()

print(quantity_subcategory)
```

```python
quantity_subcategory.plot(kind='bar')

plt.title('Quantity Sold by Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Total Quantity')
plt.xticks(rotation=45)
plt.show()
```

```python
total_sales = df['Amount'].sum()
total_profit = df['Profit'].sum()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
```

```python
profit_margin = (total_profit / total_sales) * 100

print("Profit Margin:", profit_margin, "%")
```

# Business Insights

1. The year 2022 recorded the highest total sales.
2. December recorded the highest monthly sales.
3. Office Supplies generated the highest total sales among all product categories.
4. Office Supplies generated the highest total profit among all product categories.
5. Markers generated the highest total profit among all sub-categories.
6. New York generated the highest total sales among all states.
7. Florida generated the highest total profit among all states.
8. Debit Card was the most commonly used payment mode based on total sales.
9. Cory Evans was the top customer based on total sales amount.
10. The business generated total sales of 6,182,639 and total profit of 1,610,697, resulting in an overall profit margin of approximately 26.05%.
